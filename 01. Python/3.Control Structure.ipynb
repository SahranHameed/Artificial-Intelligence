```python
# IF-ELSE Statement
# Get user input
age = int(input("Enter Your Age: "))

# Check if age is 18 or above
if(age==18):
    print("Adult")

# Otherwise
else:
    print("Children")
```

```python
# Check Even or Odd Number
num = int(input("Enter number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Check Positive, Negative or Zero
num = int(input("Enter number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
#IF–ELIF–ELSE Statement (Age Classification)
age = int(input("Enter Your Age: "))

# Check if age is less than 18
if(age<18):
    print("Children")
    
# Check if age is between 18 and 34
elif(age<35 and age>18):
    print("Adult")

# Check if age is between 35 and 44
elif(age<45 and age>35):
    print("Citizen")

# If age is 45 or above
else:
    print("Senior Citizen")
    
```

```python
# For Loop Example (List)
numbers=[15,21,38,44,56,64,79]
for temp in numbers:
    print(temp)
```

```python
# Find Maximum Number in a List
numbers = [10,25,40,65,90,100]
print(max(numbers))
```

```python
# IF Statement Using List with FOR Loop
numbers=[15,23,43,67,34,79,10]
for age in numbers:
    if(age<18):
        print("Children")
    
    elif(age<35 and age>18):
        print("Adult")
    
    elif(age<45 and age>35):
        print("Citizen")
        
    else:
        print("Senior Citizen")
    
```
